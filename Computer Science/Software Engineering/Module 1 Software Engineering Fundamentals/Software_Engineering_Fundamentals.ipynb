### **The Software Development Lifecycle (SDLC)**

The SDLC is a systematic framework covering how software is planned, built, tested, and released into use. SDLC models vary in how they are organised and what they prioritise, but a common set of underlying activities recurs across nearly all of them:

| Activity | Description |
|---|---|
| **Requirements** | Establish what the system needs to do, drawing on the needs of users, stakeholders, and regulatory bodies |
| **Design and Implementation** | Determine the system's structure, then produce the code and bring the components together |
| **Verification and Validation** | *Verification* asks whether the system is being built correctly (via testing and reviews); *validation* asks whether the correct system is being built (via stakeholder review) |
| **Maintenance** | Correct defects, adjust to changing environments, and add capability once the system is live |

### **SDLC Models Overview**

No process suits every project. Selection depends on how well requirements are understood, their likely stability, team size, risk tolerance, and regulatory context.

| Model | Approach | Best For |
|---|---|---|
| **Plan-driven (Waterfall)** | Sequential phases, each finished before the next | Stable, well-understood requirements |
| **Incremental** | Delivery in functional slices | Partial requirements; early delivery needed |
| **Agile** | Iterative; change is expected | Evolving requirements; fast feedback |
| **Open Source** | Community-driven, distributed contributions | Widely used tools and libraries |

### **Waterfall versus Incremental**

Waterfall was introduced by Winston Royce in 1970, though his paper actually presented it as flawed (Royce, 1970). It assumes requirements can be fully specified upfront. Incremental development answers Waterfall's worst flaw: the team builds and ships slices independently, core functionality first, with stakeholder feedback shaping later increments. It is the conceptual basis of Agile, but works alongside plan-driven approaches too.



| | Waterfall | Incremental |
|---|---|---|
| **Structure** | Strict sequence; no phase starts until the last ends | Independent slices, each designed, built, and delivered on its own |
| **Strengths** | Clear milestones and deliverables; easy to manage and document; suits stable domains (embedded systems, regulated government contracts) | Working software arrives early and can be redirected by real use; core features usable while secondary ones are built; a cancelled or cut project still leaves something working |
| **Weaknesses** | Requirements rarely stay stable; late-found errors are costly; nothing usable until the end; poor fit for high uncertainty | Partitioning into coherent slices takes careful planning; architecture must absorb later increments without rework; awkward for fixed-price contracts with undefined scope |

#### **The Moving Target Problem**

Requirements change,  the *moving target problem*. Legitimate drivers: users discover needs once they see early versions; the business environment shifts (market, regulation, competition); new technology enables new approaches; stakeholder compromises evolve.

| Manifestation | What Happens | Why It Matters |
|---|---|---|
| **Feature creep** | Requirements accumulate one small, reasonable addition at a time until scope far exceeds the original plan | Among the leading causes of project overruns |
| **Regression risk** | New features or bug fixes silently break working functionality | Undetected regressions reach users, as in the CBA case, where an update broke existing behaviour unnoticed |

Managing this needs processes that absorb change while protecting what already works: automated testing, disciplined change management, short feedback cycles.

#### **Limitations of Documentation-Driven Development**

Writing exhaustive specifications for client sign-off before development is the instinctive Waterfall response to changing requirements. Its limits are well documented.

| Affected Party | Limitation |
|---|---|
| **Clients** | Requirements documents are technical artefacts non-technical stakeholders often cannot meaningfully assess, a 200-page spec may be signed without real understanding, leading to disappointment on delivery |
| **Developers** | Natural language is imprecise, so written requirements are inevitably ambiguous; two developers reading the same line build two different things |
| **The project** | Documentation goes stale the moment implementation starts, a spec written for an 18-month project rarely describes the system that ships |

Documentation is not the problem; documentation *alone* is. This insight drove Agile's preference for working software and customer collaboration over comprehensive documentation.

### **Agile Software Development**

Agile is a family of approaches, not one methodology, bound together by the Agile Manifesto, written in 2001 by seventeen practitioners who had grown frustrated with heavyweight, documentation-driven processes. Its central claim: requirements and solutions emerge through collaboration, and responding to change beats sticking to a plan.

### **The Agile Manifesto**

Four core values, each a preference rather than an absolute:

| We value… | …over |
|---|---|
| Individuals and interactions | Processes and tools |
| Working software | Comprehensive documentation |
| Customer collaboration | Contract negotiation |
| Responding to change | Following a plan |

Work proceeds in short cycles, iterations or sprints, usually 1–4 weeks, each yielding a working, tested increment. Stakeholders review it, and their feedback shapes the next cycle.

| Key Principle | Meaning |
|---|---|
| Deliver frequently | Working software in weeks, not months |
| Welcome change | Even late in development |
| Collaborate daily | Business people and developers work together |
| Keep it simple | Maximise the work *not* done |

The values and principles are deliberately abstract, they name the goal, not the mechanics. Frameworks supply the mechanics. The two dominant ones: **Scrum**, a structured sprint cycle with defined roles and ceremonies, and **Kanban**, a continuous flow-based approach with fewer fixed rules.

#### **Scrum**

The most widely adopted Agile framework (Schwaber & Sutherland, 2020), defined by three sets of elements:

| Category | Element | Description |
|---|---|---|
| **Roles** | Product Owner | Represents stakeholders; owns and prioritises the product backlog |
| | Scrum Master | Facilitates the process, clears impediments, coaches the team |
| | Development Team | Self-organising group that delivers the increment |
| **Events** | Sprint | Time-boxed iteration of 1–4 weeks |
| | Sprint Planning | Team selects backlog items and plans the sprint |
| | Daily Scrum | 15-minute standup to synchronise and surface blockers |
| | Sprint Review | Increment demonstrated to stakeholders |
| | Sprint Retrospective | Team reflects on process and identifies improvements |
| **Artefacts** | Product Backlog | Ordered list of everything the product might need |
| | Sprint Backlog | Items chosen for the current sprint, plus the delivery plan |
| | Increment | Sum of all completed backlog items at sprint end |

#### **Kanban**

Adapted from Toyota's manufacturing system by David Anderson (Anderson, 2010), Kanban is flow-based: visualise the work, cap work in progress, and improve continuously. A board shows work as cards moving through columns.

| Practice | What It Involves |
|---|---|
| Visualise the workflow | Make all work and its current status visible |
| Limit WIP | Prevent overload, finish before starting more |
| Manage flow | Track cycle time and throughput; find bottlenecks |
| Improve collaboratively | Let data drive continuous improvement |

Kanban fits teams facing highly variable incoming work, support and maintenance, for instance, or anyone wanting a lighter alternative to Scrum's ceremonies.

### **Rapid Prototyping**

Agile softens Waterfall's rigidity but still assumes stakeholders can articulate their wants, at least enough to write user stories and rank a backlog. In reality, users often cannot describe their needs until they have something concrete to react to. Sprint reviews help, yet a four-week sprint gives a team ample time to build in the wrong direction on unclear requirements. Agile lowers the cost of late change; it does not remove initial misunderstanding. Rapid prototyping, usable with any process model, closes that gap.

Rapid prototyping means building a quick, rough version of the system (or a key part) to gather feedback before committing to full implementation. A prototype is not a product; it is a communication and learning tool.

| Type | How It Works | Purpose |
|---|---|---|
| **Throwaway** | Built quickly, shown to stakeholders, then discarded; code is not production-quality | Validate understanding |
| **Evolutionary** | Built incrementally and progressively refined | Becomes the final system |

The value is simple: users react far more effectively to something they can see and use than to something they can only read about. Misunderstandings surface early, when correcting them is cheap, rather than late, when it is not.

### **Open Source Development**

Open source development makes source code publicly available and builds it collaboratively through a distributed community. Anyone may inspect, use, modify, and distribute the software within the terms of its licence.

The modern movement began with Richard Stallman's GNU project (1983) and accelerated sharply after Linus Torvalds created the Linux kernel in 1991. Open source now underpins much of the internet's infrastructure, web servers (Apache, Nginx), programming languages (Python, Ruby), and mobile operating systems (Android, built on the Linux kernel).

| Characteristic | Description |
|---|---|
| **Community-driven** | Contributors come from individuals and organisations with mixed motives, learning, reputation, commercial interest, ideology |
| **Distributed** | Contributors work asynchronously from anywhere in the world |
| **Transparent** | Code, issues, and discussions are publicly visible and open to review |
| **Release early, release often** | Rapid iteration and public feedback stand in for formal specification |

The model poses genuine engineering questions: how is quality maintained when anyone can contribute, and how are architectural decisions made by committee? Answering them produced code review workflows, continuous integration, and community governance models, practices now standard in commercial development too.
