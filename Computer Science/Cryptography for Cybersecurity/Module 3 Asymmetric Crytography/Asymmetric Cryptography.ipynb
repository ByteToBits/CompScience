### **Introduction to Asymmetric Cryptography**

**Symmetric Cryptography** is fast and efficient, but it depends completely on keeping a **Shared Secret Key** safe. This creates two main problems:

- **Key Exchange**: both parties need the same key before they can communicate
  - Two people on opposite sides of the world have to find a secure way to share the key first
  - This was the hardest problem to solve
- **Non-Repudiation**: there's no proof of who actually sent a message
  - Both parties have the same key, so either of them could have created the message
  - The sender can deny sending it

Symmetric ciphers are built from **Substitution and Permutation**, and neither can solve these problems. The solution was **Asymmetric Cryptography**, also called **Public-Key Cryptography**.

#### **How Asymmetric Cryptography Works**

Instead of one shared key, each user has a **Pair of Related Keys**:

- **Public Key** ($PU$): shared with anyone
- **Private Key** ($PR$): kept secret by its owner

The key pair is generated with a **one-way function**: easy to compute in one direction, but infeasible to reverse. This is where the **hard problems** from lesson 3 come in, such as integer factorisation for RSA. What one key does, **only the other key** can undo.

#### **Encryption with the Public Key: Confidentiality**

Bob wants to send Alice a secret message. He encrypts it with **Alice's public key**, so **only Alice** can decrypt it with her private key.

```mermaid
flowchart LR
    X1["Plaintext X"] --> E1["Bob encrypts<br>(e.g. RSA)"] -- "Y = E(PU<sub>a</sub>, X)" --> D1["Alice decrypts"] --> X2["Plaintext X"]
    K1(["Alice's public key PU<sub>a</sub>"]) --> E1
    K2(["Alice's private key PR<sub>a</sub>"]) --> D1
```

- **Confidentiality**:
  - Bob encrypts with Alice's **public** key: $Y = E(PU_a, X)$
  - Alice decrypts with her **private** key: $X = D(PR_a, Y)$
  - Anyone can encrypt messages to Alice, but only Alice can read them
  - **Solves key exchange**: no secret needs to be shared in advance
  - Each user keeps a **public key ring**: a collection of other people's public keys

#### **Encryption with the Private Key: Authentication**

Bob encrypts a message with **his own private key**. Anyone can decrypt it with Bob's public key, so it **isn't secret**. But since only Bob has his private key, it proves the message **came from Bob**.

```mermaid
flowchart LR
    X1["Plaintext X"] --> E1["Bob encrypts<br>(e.g. RSA)"] -- "Y = E(PR<sub>b</sub>, X)" --> D1["Alice decrypts"] --> X2["Plaintext X"]
    K1(["Bob's private key PR<sub>b</sub>"]) --> E1
    K2(["Bob's public key PU<sub>b</sub>"]) --> D1
```

- **Authentication**:
  - Bob encrypts with his **private** key: $Y = E(PR_b, X)$
  - Alice decrypts with Bob's **public** key: $X = D(PU_b, Y)$
  - Anyone can decrypt it, so it provides **no confidentiality**
  - Only Bob could have created it, so it proves who sent it and provides **non-repudiation**
  - This is the idea behind **digital signatures**

| Encrypt With        | Decrypt With        | Who Can Read It?  | Provides                           |
|---------------------|---------------------|-------------------|------------------------------------|
| Receiver's public key | Receiver's private key | Only the receiver | Confidentiality                  |
| Sender's private key  | Sender's public key    | Anyone            | Authentication, non-repudiation  |

#### **Requirements for a Public-Key Algorithm**

- **Requirements**:
  - **Key generation**: user $b$ can easily generate a key pair $(PU_b, PR_b)$
  - **Encryption**: a sender who knows $PU_b$ can easily encrypt a message $M$: $C = E(PU_b, M)$
  - **Decryption**: $b$ can easily decrypt using $PR_b$: $M = D(PR_b, C) = D[PR_b, E(PU_b, M)]$
  - **Private key stays private**: knowing the algorithm and $PU_b$, it's infeasible to work out $PR_b$
  - **Message stays secret**: knowing $PU_b$ and $C$, it's infeasible to recover $M$
  - **Either order (optional)**: either key can encrypt and the other decrypts
    - $M = D[PU_b, E(PR_b, M)] = D[PR_b, E(PU_b, M)]$
    - True for RSA, but not every public-key algorithm supports this

#### **Components of a Public-Key Cryptosystem**

- **Plaintext**: the message to be sent
- **Encryption algorithm**: transforms the plaintext using one of the keys
- **Public key**: the half of the key pair that is shared openly
- **Private key**: the half of the key pair kept secret by its owner
  - Not the same as a **secret key**, which is the shared key in symmetric cryptography
- **Ciphertext**: the output of the encryption algorithm
- **Decryption algorithm**: transforms the ciphertext back into plaintext using the other key

#### **Symmetric vs Asymmetric Cryptography**

| Feature           | Symmetric                          | Asymmetric                               |
|-------------------|------------------------------------|------------------------------------------|
| Keys              | One shared secret key              | Public and private key pair              |
| Based on          | Substitution and permutation       | Hard mathematical problems               |
| Key exchange      | Must be done securely in advance   | Public key can be shared openly          |
| Non-repudiation   | No                                 | Yes, with the private key                |
| Speed             | Fast                               | Much slower                              |
| Examples          | AES, DES, ChaCha20                 | RSA, Diffie-Hellman, ECC                 |

In practice, the two are **used together**. Asymmetric cryptography safely sets up a shared key, then fast symmetric encryption (like AES) protects the actual data. This is how HTTPS and VPNs work.

### **Applications of Asymmetric Cryptosystems**

#### **The Key Management Problem**

With symmetric cryptography, **every pair of users** who want to talk privately needs **their own shared secret key**. That's fine for a few users, but the number of keys grows very quickly as more users join.

With asymmetric cryptography, each user only needs **One Key Pair**. Anyone can use the public key to encrypt a message to its owner, but only the owner's private key can decrypt it. Even an attacker who knows the public key used to encrypt a message can't decrypt it.

$$\text{Symmetric keys} = \frac{n(n - 1)}{2} \qquad \text{Asymmetric key pairs} = n$$

| Users ($n$) | Symmetric Keys | Asymmetric Key Pairs |
|-------------|----------------|----------------------|
| 2           | 1              | 2                    |
| 10          | 45             | 10                   |
| 100         | 4,950          | 100                  |
| 1,000       | 499,500        | 1,000                |

#### **Efficiency**

Asymmetric cryptography is **much slower** than symmetric cryptography. It relies on heavy maths, like modular exponentiation with very large numbers, and needs **much longer keys** for the same level of security.

| Security Level | Symmetric (e.g. AES) | RSA         | Elliptic Curve (ECC) |
|----------------|----------------------|-------------|----------------------|
| 112 bits       | 112 (3DES)           | 2048 bits   | 224 bits             |
| 128 bits       | 128                  | 3072 bits   | 256 bits             |
| 192 bits       | 192                  | 7680 bits   | 384 bits             |
| 256 bits       | 256                  | 15360 bits  | 512 bits             |

Despite being slower, asymmetric cryptography is widely used for three things.

#### **1. Encryption and Decryption**

Encrypting a message with the **recipient's public key** means only the recipient can read it. This is useful when the message has to travel over an untrusted channel, like email.

- **Encryption and Decryption**:
  - Gives **confidentiality** for a specific recipient
  - Too slow for large amounts of data
  - In practice, the message is encrypted with a random **symmetric key**, and only that small key is encrypted with the recipient's public key
  - This is how encrypted email (PGP, S/MIME) works

#### **2. Digital Signatures**

Probably the best-known use. The sender **signs** a message with their **private key**, and anyone can **verify** it with the sender's **public key**. In practice, a short **hash** of the message is signed rather than the whole message.

```mermaid
flowchart LR
    M["Message"] --> H["Hash"] --> S["Sign with<br>sender's private key"] --> SIG["Message + signature"]
    SIG --> V["Verify with<br>sender's public key"] --> OK["Valid: sent by the sender<br>and not changed"]
```

- **Digital Signatures**:
  - **Authentication**: the receiver can be confident the message came from the claimed sender
  - **Non-repudiation**: the sender can't deny sending it, since only they have the private key
  - **Integrity**: any change to the message makes the signature invalid
  - No confidentiality: the message itself is not hidden

#### **3. Key Exchange**

The **most common** use. Symmetric ciphers are fast, but sharing their key securely is hard. Asymmetric cryptography solves this by **setting up the symmetric key**, and then the fast symmetric cipher protects the actual data. This is called **Hybrid Encryption**.

```mermaid
sequenceDiagram
    participant A as Alice
    participant B as Bob
    Note over A,B: Step 1 - Asymmetric (slow, used once)
    A->>B: Set up a shared secret key K using RSA or Diffie-Hellman
    Note over A,B: Step 2 - Symmetric (fast, used for all data)
    A->>B: Data encrypted with AES using K
    B->>A: Data encrypted with AES using K
```

- **Two Ways to Exchange a Key**:
  - **Key transport**: one side picks a random key, encrypts it with the other side's **public key**, and sends it (e.g. RSA)
  - **Key agreement**: both sides exchange public values and each **calculates the same key** on their own, so the key is never sent at all (e.g. Diffie-Hellman)
  - Used by HTTPS (TLS), VPNs and messaging apps

| Application             | Key Used to Encrypt or Sign | Key Used to Decrypt or Verify | Provides                                   | Example Algorithms |
|-------------------------|-----------------------------|-------------------------------|--------------------------------------------|--------------------|
| Encryption / decryption | Receiver's public key       | Receiver's private key        | Confidentiality                            | RSA                |
| Digital signature       | Sender's private key        | Sender's public key           | Authentication, integrity, non-repudiation | RSA, DSA, ECDSA    |
| Key exchange            | Both parties' key pairs     | Both parties' key pairs       | A shared symmetric key                     | RSA, Diffie-Hellman |

### **Trap-Door One-Way Functions**

The requirements for a public-key algorithm are hard to meet. Since public-key cryptography was introduced in the **late 1970s**, only a **handful of algorithms** (such as RSA, Diffie-Hellman and elliptic curve cryptography) have become widely used. The most technical requirement is the **Trap-Door One-Way Function**.

The term **one-way function** has several meanings in cryptography, depending on the context. It comes up again with **hash functions** in the lesson on digital signatures, where the function can't be reversed at all. Here it means something slightly different.

#### **One-Way Functions**

A **One-Way Function** is **easy to compute** in one direction but **infeasible to reverse**. For public-key cryptography, the function must also be **invertible**: every output $Y$ comes from exactly one input $X$, so an inverse exists, even if it can't be found.

$$Y = f(X) \quad \longrightarrow \quad \text{easy to calculate } Y \text{ given } X$$

$$X = f^{-1}(Y) \quad \longrightarrow \quad \text{infeasible to calculate } X \text{ given } Y$$

- **Easy vs Infeasible**: measured against the input size $n$ (in bits)
  - **Easy**: solvable in **polynomial time**, e.g. $n^2$ or $n^3$ steps
  - **Infeasible**: no **known** algorithm solves it in polynomial time, e.g. it takes about $2^n$ steps
  - With $n = 2048$, $2^n$ steps would take far longer than the age of the universe

#### **Trap-Door One-Way Functions**

A **Trap-Door One-Way Function** is a one-way function with a **secret shortcut**, the **trapdoor** $k$. Without $k$, reversing the function is infeasible. With $k$, it becomes easy (polynomial time).

```mermaid
flowchart LR
    X["X"] -- "Easy: Y = f(X)<br>using the public key" --> Y["Y"]
    Y -- "Easy with trapdoor k" --> X2["X"]
    Y -. "Infeasible without k" .-> Q["?"]
```

It is a family of invertible functions $f_k$ where:

$$Y = f_k(X) \quad \longrightarrow \quad \text{easy, if } k \text{ and } X \text{ are known}$$

$$X = f_k^{-1}(Y) \quad \longrightarrow \quad \text{easy, if } k \text{ and } Y \text{ are known}$$

$$X = f_k^{-1}(Y) \quad \longrightarrow \quad \text{infeasible, if } Y \text{ is known but } k \text{ is not}$$

- **Mapping to Public-Key Cryptography**:
  - **Computing $Y = f_k(X)$** is encryption, and only needs the **public key**
  - **The trapdoor $k$** is the **private key**
  - **Computing $X = f_k^{-1}(Y)$** is decryption, which is only easy for the private key holder
  - Every public-key algorithm needs its own trapdoor function, and very few good ones are known

#### **Trapdoor Execution: RSA with Small Numbers**

RSA's trapdoor function is $f(X) = X^e \bmod n$. Reversing it means finding $e$-th roots mod $n$, which is infeasible unless you know the **prime factors** of $n$.

| Step | Who Can Do It | Working | Result |
|------|---------------|---------|--------|
| Choose primes (trapdoor) | Key owner only | $p = 5$, $q = 11$ | Secret |
| Public key | Anyone | $n = pq = 55$, $e = 3$ | $(55, 3)$ |
| Encrypt $X = 7$ (easy) | Anyone | $7^3 \bmod 55 = 343 \bmod 55$ | $Y = 13$ |
| Find private key (needs trapdoor) | Key owner only | $\phi(55) = 4 \times 10 = 40$, $d = 3^{-1} \bmod 40$ | $d = 27$ |
| Decrypt $Y = 13$ (easy with $d$) | Key owner only | $13^{27} \bmod 55$ | $X = 7$ ✓ |
| Decrypt without the trapdoor | Attacker | Must factor $n$ to find $\phi(n)$ | Infeasible for large $n$ |
