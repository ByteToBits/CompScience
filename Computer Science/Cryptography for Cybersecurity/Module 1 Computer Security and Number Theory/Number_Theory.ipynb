### **Mathematical Foundations in Cryptography**

**Finite Fields** are one of the most important concepts in cryptography. Almost all cryptographic algorithms are built on them, which lets us use proven mathematical theorems to **Justify that an Algorithm is Sound and Correct**.

Maths that is slow for humans is fast for computers, so as computers get faster, cryptography uses **Larger and Larger Numbers**. But computation is not free, and bigger numbers mean more cycles. So cryptography uses large numbers that can be **Reduced to be as Small as Possible**, cutting down computation without breaking the underlying maths.

- **Why Finite Fields Matter**:
  - Basis for the design, operation and implementation of most cryptographic algorithms
  - Let us rely on proven theorems to show algorithms are correct

- **The Size Trade-off**:
  - Faster computers mean cryptography needs larger numbers to stay secure
  - Larger numbers need more computation cycles
  - Solution: use large numbers that can be reduced, keeping the maths valid while saving computation

### **The Division Algorithm**

**Division** is one of the most common ways to reduce large numbers. It is basic arithmetic, but it plays a **Central Role in Cryptographic Algorithms**.

#### **Divisibility**

The notation $b \mid a$ means **$b$ Divides $a$**, where $a$ is an integer and $b$ is a nonzero integer. In other words, $a = mb$ for some integer $m$, so dividing $a$ by $b$ leaves no remainder.

- **Properties of Divisibility**:
  - If $a \mid 1$, then $a = \pm 1$
  - If $a \mid b$ and $b \mid a$, then $a = \pm b$
  - $b \mid 0$ for any $b \neq 0$
  - If $a \mid b$ and $b \mid c$, then $a \mid c$
  - If $b \mid g$ and $b \mid h$, then $b \mid (mg + nh)$ for any integers $m$ and $n$

| Expression   | True? | Reason                       |
|--------------|-------|------------------------------|
| $13 \mid 182$ | Yes   | $182 = 14 \times 13$         |
| $-5 \mid 30$  | Yes   | $30 = -6 \times -5$          |
| $17 \mid 0$   | Yes   | $0 = 0 \times 17$            |
| $4 \mid 10$   | No    | $10 = 2 \times 4 + 2$, remainder is not 0 |

#### **The Division Algorithm**

The **Division Theorem**, usually called the **Division Algorithm**, says that dividing a non-negative integer $a$ by a positive integer $n$ always gives a **Quotient** $q$ and a **Remainder** $r$:

$$a = qn + r, \quad 0 \le r < n, \quad q = \lfloor a/n \rfloor$$

- **Division Algorithm**: $a = qn + r$
  - $a$: the number being divided (dividend)
  - $n$: the number dividing it (divisor)
  - $q = \lfloor a/n \rfloor$: the quotient, rounded down
  - $r$: the remainder, always between $0$ and $n - 1$

| $a$ | $n$ | $q = \lfloor a/n \rfloor$ | $r = a - qn$ | Check: $qn + r$ |
|-----|-----|---------------------------|--------------|-----------------|
| 11  | 7   | 1                         | 4            | $1 \times 7 + 4 = 11$  |
| 25  | 5   | 5                         | 0            | $5 \times 5 + 0 = 25$  |
| 3   | 8   | 0                         | 3            | $0 \times 8 + 3 = 3$   |
| 100 | 9   | 11                        | 1            | $11 \times 9 + 1 = 100$ |

### **The Euclidean Algorithm**

The **Euclidean Algorithm** finds the **Greatest Common Divisor (GCD)** of two positive integers. Cryptography often needs large numbers that are prime or **Relatively Prime**, which is a requirement for most cryptographic algorithms. Two numbers are relatively prime if their **Only Common Divisor is 1**, meaning $\gcd(a, b) = 1$.

#### **Greatest Common Divisor**

The GCD of $a$ and $b$ is the largest integer $k$ that divides both of them:

$$\gcd(a, b) = \max[k], \text{ such that } k \mid a \text{ and } k \mid b$$

In cryptography, the GCD is always kept as a **Positive Integer**.

- **GCD Properties**:
  - $\gcd(a, b)$ is the largest $k$ where $k \mid a$ and $k \mid b$
  - $\gcd(a, 0) = |a|$ for any nonzero integer $a$
  - $a$ and $b$ are relatively prime if $\gcd(a, b) = 1$

#### **How the Algorithm Works**

The Euclidean algorithm uses the **Division Algorithm Repeatedly**. Each step divides the previous divisor by the previous remainder, giving a new, smaller remainder. This repeats until the **Remainder Becomes 0**, and the last nonzero remainder is the GCD.

It assumes $a \ge b > 0$. The order doesn't really matter, but the process starts by dividing the larger number by the smaller one.

$$
\begin{aligned}
a &= q_1 b + r_1 && 0 < r_1 < b \\
b &= q_2 r_1 + r_2 && 0 < r_2 < r_1 \\
r_1 &= q_3 r_2 + r_3 && 0 < r_3 < r_2 \\
&\ \ \vdots \\
r_{n-2} &= q_n r_{n-1} + r_n && 0 < r_n < r_{n-1} \\
r_{n-1} &= q_{n+1} r_n + 0 \\
d &= \gcd(a, b) = r_n
\end{aligned}
$$

- **Euclidean Algorithm**: Repeated division until the remainder is 0
  - Start by dividing the larger number by the smaller one
  - Each step: the old divisor becomes the new dividend, the old remainder becomes the new divisor
  - Remainders get smaller every step, so the algorithm always finishes
  - Last nonzero remainder $r_n$ is the GCD

- **Why It Works**: $\gcd(a, b) = \gcd(b, r)$
  - Any number that divides $a$ and $b$ also divides $r = a - qb$
  - So the GCD stays the same at every step, while the numbers get smaller

#### **Euclidean Algorithm Execution for $\gcd(1160718174, 316258250)$**

| Step | Division                                    | Quotient | Remainder   | Reduced To                     |
|------|---------------------------------------------|----------|-------------|--------------------------------|
| 1    | $1160718174 = 3 \times 316258250 + 211943424$ | 3        | 211943424   | $\gcd(316258250, 211943424)$   |

### **Modular Arithmetic**

**Modular Arithmetic** works with remainders. It keeps numbers within a fixed range, which is why it is so **Important for Cryptographic Algorithms**.

#### **The Modulus**

From the division algorithm, any integer $a$ can be written as:

$$a = qn + r, \quad 0 \le r < n, \quad q = \lfloor a/n \rfloor$$

$$a = \lfloor a/n \rfloor \cdot n + (a \bmod n)$$

$a \bmod n$ is the **Remainder When $a$ is Divided by $n$**. The modulus is always written as the **Smallest Non-Negative Number**, so for negative numbers we keep adding $n$ until the result lands between $0$ and $n - 1$.

- **Modulus**: $a \bmod n = r$
  - $r$ is the remainder of $a \div n$
  - Always in the range $0 \le r < n$
  - Negative numbers are reduced to the smallest non-negative remainder

| $a$  | $n$ | $q = \lfloor a/n \rfloor$ | $a \bmod n$ | Check: $qn + r$        |
|------|-----|---------------------------|-------------|------------------------|
| 11   | 7   | 1                         | 4           | $1 \times 7 + 4 = 11$  |
| -11  | 7   | -2                        | 3           | $-2 \times 7 + 3 = -11$ |
| 25   | 5   | 5                         | 0           | $5 \times 5 + 0 = 25$  |

#### **Congruence**

The modulus can be used in two ways. As a **Binary Operator**, $a \bmod n$ gives a remainder. As a **Congruence Relation**, it tells us whether two integers are **Equivalent**:

$$a \equiv b \pmod{n} \quad \text{if} \quad (a \bmod n) = (b \bmod n)$$

For example, $11 \equiv 4 \pmod{7}$ and $-11 \equiv 3 \pmod{7}$.

- **Properties of Congruence**:
  - $a \equiv b \pmod{n}$ if $n \mid (a - b)$
  - $a \equiv b \pmod{n} \Rightarrow b \equiv a \pmod{n}$
  - $a \equiv b \pmod{n}$ and $b \equiv c \pmod{n} \Rightarrow a \equiv c \pmod{n}$

#### **Modular Arithmetic Operations**

Modulo $n$ reduces any integer into the set $\{0, 1, 2, \ldots, n - 1\}$, so we can do arithmetic within that set. The key idea is that **You Can Reduce Before or After** an operation and get the same answer.

- **Modular Addition**: $[(a \bmod n) + (b \bmod n)] \bmod n = (a + b) \bmod n$
- **Modular Subtraction**: $[(a \bmod n) - (b \bmod n)] \bmod n = (a - b) \bmod n$
- **Modular Multiplication**: $[(a \bmod n) \cdot (b \bmod n)] \bmod n = (a \cdot b) \bmod n$

All three proofs follow the same steps. Write $a$ and $b$ using the division algorithm:

$$a = nq_1 + r_1, \quad a \bmod n = r_1 \qquad b = nq_2 + r_2, \quad b \bmod n = r_2$$

| Operation      | Expand $a \circ b$                          | Multiples of $n$ Removed | Result                          |
|----------------|---------------------------------------------|--------------------------|---------------------------------|
| Addition       | $n(q_1 + q_2) + r_1 + r_2$                  | $n(q_1 + q_2)$           | $(r_1 + r_2) \bmod n$ ✓         |
| Subtraction    | $n(q_1 - q_2) + r_1 - r_2$                  | $n(q_1 - q_2)$           | $(r_1 - r_2) \bmod n$ ✓         |
| Multiplication | $n^2 q_1 q_2 + nq_2 r_1 + nq_1 r_2 + r_1 r_2$ | $n(nq_1q_2 + q_2r_1 + q_1r_2)$ | $(r_1 \cdot r_2) \bmod n$ ✓ |

In each case, every term containing $n$ disappears under $\bmod n$. Only the remainders are left, which is exactly the left-hand side, so **LHS = RHS**.

#### **Modular Exponentiation**

Modular exponentiation is **Repeated Modular Multiplication**. The problem is that powers grow huge very quickly, especially with cryptographic numbers. The fix is to **Reduce After Every Multiplication** so the numbers stay small:

$$a^2 \bmod c = (a \cdot a) \bmod c = [(a \bmod c) \cdot (a \bmod c)] \bmod c$$

- **Repeated Squaring**: Square the previous result and reduce each time
  - $a^{2k} \bmod n = [(a^k \bmod n) \cdot (a^k \bmod n)] \bmod n$
  - Numbers never get bigger than $(n - 1)^2$ before reducing
  - Works directly when the exponent is a power of 2

- **Any Exponent**: Use the binary form of the exponent
  - Write the exponent as a sum of powers of 2
  - Compute each needed power with repeated squaring
  - Multiply those results together, reducing after each multiplication

#### **Modular Exponentiation Execution for $7^{256} \bmod 13$**

$256 = 2^8$, so we just square 8 times.

| Step | Power     | Calculation                           | Result |
|------|-----------|---------------------------------------|--------|
| 1    | $7^{1}$   | $7 \bmod 13$                          | 7      |
| 2    | $7^{2}$   | $(7 \times 7) \bmod 13 = 49 \bmod 13$ | 10     |
| 3    | $7^{4}$   | $(10 \times 10) \bmod 13 = 100 \bmod 13$ | 9   |
| 4    | $7^{8}$   | $(9 \times 9) \bmod 13 = 81 \bmod 13$ | 3      |

### **Fermat's Little Theorem**

**Fermat's Little Theorem** is an important result in cryptography. It says that for a **Prime Number $p$** and a positive integer **$a$ Not Divisible by $p$**:

$$a^{p-1} \equiv 1 \pmod{p}$$

Multiplying both sides by $a$ gives another common form:

$$a^{p} \equiv a \pmod{p}$$

Since $\gcd(a, p) = 1$, we can divide both sides by $a$ to get back to the first form. The second form also holds when $p \mid a$, because both sides are then $0$.

- **Fermat's Little Theorem**:
  - $a^{p-1} \equiv 1 \pmod{p}$ when $p$ is prime and $p \nmid a$
  - $a^{p} \equiv a \pmod{p}$ for any integer $a$ when $p$ is prime
  - Useful for shrinking large exponents: exponents can be reduced modulo $p - 1$

| $a$ | $p$ | $a^{p-1}$       | $a^{p-1} \bmod p$ |
|-----|-----|-----------------|-------------------|
| 2   | 5   | $2^4 = 16$      | 1                 |
| 3   | 7   | $3^6 = 729$     | 1                 |
| 4   | 11  | $4^{10} = 1048576$ | 1              |

#### **Proof**

The idea is that multiplying every number from $1$ to $p - 1$ by $a$ just **Shuffles Them Around** modulo $p$. So both lists have the same product.

Let $S$ be the set of positive integers less than $p$:

$$S = \{1, 2, 3, \ldots, p - 1\}$$

Multiply each element by $a$ and reduce modulo $p$:

$$S' = \{1a, 2a, 3a, \ldots, (p - 1)a\} \pmod{p}$$

- **No element of $S'$ is 0**: $p$ doesn't divide $a$ or any of $1, \ldots, p - 1$, so $p$ doesn't divide their product
- **All elements of $S'$ are distinct**:
  - Suppose $ai \equiv aj \pmod{p}$
  - Since $\gcd(a, p) = 1$, we can cancel $a$ to get $i \equiv j \pmod{p}$
  - Since $1 \le i, j \le p - 1$, this means $i = j$
- **So $S'$ is a permutation of $S$**: same $p - 1$ values, just in a different order

Because $S'$ and $S$ contain the same values, their products are equal modulo $p$:

$$1a \cdot 2a \cdot 3a \cdot \ldots \cdot (p - 1)a \equiv 1 \cdot 2 \cdot 3 \cdot \ldots \cdot (p - 1) \pmod{p}$$

$$a^{p-1}(p - 1)! \equiv (p - 1)! \pmod{p}$$

$(p - 1)!$ has no factor of $p$, so it is relatively prime to $p$ and can be cancelled from both sides:

$$a^{p-1} \equiv 1 \pmod{p} \qquad \blacksquare$$

#### **Proof Execution for $a = 3$, $p = 7$**

| $i$ | $3i$ | $3i \bmod 7$ |
|-----|------|--------------|
| 1   | 3    | 3            |
| 2   | 6    | 6            |
| 3   | 9    | 2            |
| 4   | 12   | 5            |
| 5   | 15   | 1            |
| 6   | 18   | 4            |

Final result:
- $S = \{1, 2, 3, 4, 5, 6\}$
- $S' = \{3, 6, 2, 5, 1, 4\}$, the same values in a different order
- Product of $S$: $6! = 720$
- Product of $S'$: $3^6 \cdot 6! \equiv 6! \pmod{7}$
- Cancelling $6!$: $3^6 \equiv 1 \pmod{7}$ ✓

#### **Using the Theorem to Reduce Exponents**

Since $a^{p-1} \equiv 1 \pmod{p}$, any multiple of $p - 1$ in the exponent can be removed.

| Expression          | Reduce Exponent mod $(p - 1)$       | Simplified      | Result |
|---------------------|-------------------------------------|-----------------|--------|
| $3^{100} \bmod 7$   | $100 = 16 \times 6 + 4$             | $3^{4} \bmod 7 = 81 \bmod 7$ | 4      |
| $2^{50} \bmod 5$    | $50 = 12 \times 4 + 2$              | $2^{2} \bmod 5 = 4 \bmod 5$  | 4      |

### **Euler's Totient Function and Euler's Theorem**

#### **Euler's Totient Function**

**Euler's Totient Function**, written $\phi(n)$ (the Greek letter **phi**), counts **How Many Positive Integers Less Than $n$ are Relatively Prime to $n$**. By convention, $\phi(1) = 1$.

A number $k$ is counted if $\gcd(k, n) = 1$, meaning $k$ and $n$ share no common factor other than 1. For example, 10 is not counted in $\phi(35)$ because 10 and 35 share the factor 5.

- **Totient Function**: $\phi(n)$ = number of integers $1 \le k < n$ with $\gcd(k, n) = 1$
  - $\phi(1) = 1$ by convention

- **For a Prime $p$**: $\phi(p) = p - 1$
  - Every number below a prime is relatively prime to it

- **For Two Distinct Primes $p$ and $q$**: $\phi(pq) = \phi(p) \cdot \phi(q) = (p - 1)(q - 1)$
  - Only works when $p \neq q$
  - This is the case used in RSA

#### **Totient Execution for $\phi(35)$**

$35 = 5 \times 7$, so any number sharing a factor of 5 or 7 with 35 is removed.

| Numbers from 1 to 34 | Removed?         | Reason        |
|----------------------|------------------|---------------|
| 5, 10, 15, 20, 25, 30 | ✗                | Multiple of 5 |
| 7, 14, 21, 28        | ✗                | Multiple of 7 |
| 1, 2, 3, 4, 6, 8, 9, 11, 12, 13, 16, 17, 18, 19, 22, 23, 24, 26, 27, 29, 31, 32, 33, 34 | ✓ Counted | No common factor |

Final counts:
- Total numbers from 1 to 34: 34
- Removed: 6 multiples of 5 + 4 multiples of 7 = 10
- $\phi(35) = 34 - 10 = 24$
- Check with the formula: $\phi(35) = (5 - 1)(7 - 1) = 4 \times 6 = 24$ ✓

| $n$ | Type                     | Formula             | $\phi(n)$ |
|-----|--------------------------|---------------------|-----------|
| 37  | Prime                    | $37 - 1$            | 36        |
| 35  | $5 \times 7$             | $(5 - 1)(7 - 1)$    | 24        |
| 21  | $3 \times 7$             | $(3 - 1)(7 - 1)$    | 12        |
| 10  | $2 \times 5$             | $(2 - 1)(5 - 1)$    | 4         |

#### **Euler's Theorem**

**Euler's Theorem** uses the totient function. For any integers $a$ and $n$ that are **Relatively Prime**:

$$a^{\phi(n)} \equiv 1 \pmod{n}$$

Multiplying both sides by $a$ gives an alternative form:

$$a^{\phi(n)+1} \equiv a \pmod{n}$$

The first form needs $\gcd(a, n) = 1$. The second form doesn't, but only when $n$ is a product of distinct primes, like $n = pq$.

- **Euler's Theorem**: $a^{\phi(n)} \equiv 1 \pmod{n}$
  - Requires $\gcd(a, n) = 1$
  - **Fermat's Little Theorem is a special case**: when $n = p$ is prime, $\phi(p) = p - 1$, giving $a^{p-1} \equiv 1 \pmod{p}$

- **Alternative Form**: $a^{\phi(n)+1} \equiv a \pmod{n}$
  - Does not require $\gcd(a, n) = 1$ when $n = pq$ with distinct primes
  - This is the property that makes RSA decryption work

| $a$ | $n$ | $\phi(n)$ | $a^{\phi(n)}$  | $a^{\phi(n)} \bmod n$ |
|-----|-----|-----------|----------------|-----------------------|
| 3   | 10  | 4         | $3^4 = 81$     | 1                     |
| 2   | 21  | 12        | $2^{12} = 4096$ | 1                    |
| 2   | 35  | 24        | $2^{24} = 16777216$ | 1                |

*The proof for this theorem is in the readings for this lesson.*

### **Extended Euclidean Algorithm**

The **Extended Euclidean Algorithm (EEA)** builds on the Euclidean algorithm. It first finds $\gcd(a, b)$, then **Works Backwards** to find two integers $x$ and $y$ such that:

$$ax + by = \gcd(a, b)$$

$x$ and $y$ are called the **Bézout Coefficients** of $a$ and $b$. The EEA is important in cryptography because it is used to find **Modular Multiplicative Inverses**.

- **Extended Euclidean Algorithm**: Finds $x$ and $y$ where $ax + by = \gcd(a, b)$
  - Step 1: Run the Euclidean algorithm to find the GCD
  - Step 2: Rewrite each step to make its remainder the subject
  - Step 3: Starting from the GCD, substitute the earlier remainders back in one at a time
  - Step 4: Collect terms until only $a$ and $b$ are left

#### **EEA Execution for $1914x + 899y = \gcd(1914, 899)$**

**Step 1: Euclidean Algorithm**

| Step | Division                    | Rearranged for Remainder        |
|------|-----------------------------|---------------------------------|
| 1    | $1914 = 2 \times 899 + 116$ | $116 = 1914 - 2 \times 899$     |
| 2    | $899 = 7 \times 116 + 87$   | $87 = 899 - 7 \times 116$       |
| 3    | $116 = 1 \times 87 + 29$    | $29 = 116 - 1 \times 87$        |
| 4    | $87 = 3 \times 29 + 0$      | Remainder is 0, stop            |

The last nonzero remainder is 29, so $\gcd(1914, 899) = 29$.

**Step 2: Work Backwards**

| Step | Substitute                   | Equation                                         | Simplified                          |
|------|------------------------------|--------------------------------------------------|-------------------------------------|
| 1    | Start from GCD               | $29 = 116 - 1 \times 87$                         | $29 = 116 - 87$                     |
| 2    | $87 = 899 - 7 \times 116$    | $29 = 116 - (899 - 7 \times 116)$                | $29 = 8 \times 116 - 1 \times 899$  |
| 3    | $116 = 1914 - 2 \times 899$  | $29 = 8 \times (1914 - 2 \times 899) - 1 \times 899$ | $29 = 8 \times 1914 - 17 \times 899$ |

Final result:
- $29 = 8 \times 1914 - 17 \times 899$
- Therefore, $x = 8$ and $y = -17$
- Check: $8 \times 1914 - 17 \times 899 = 15312 - 15283 = 29$ ✓

#### **Using the EEA to Find Modular Inverses**

If $\gcd(a, n) = 1$, the EEA gives $ax + ny = 1$. Taking both sides modulo $n$ removes the $ny$ term:

$$ax \equiv 1 \pmod{n}$$

So $x$ is the **Inverse of $a$ Modulo $n$**, written $a^{-1} \bmod n$.

- **Modular Inverse**: $a^{-1} \bmod n$ is the number $x$ where $ax \equiv 1 \pmod{n}$
  - Only exists when $\gcd(a, n) = 1$
  - If $x$ comes out negative, add $n$ to make it positive

| Find               | Euclidean Steps              | Work Backwards       | Inverse              | Check                        |
|--------------------|------------------------------|----------------------|----------------------|------------------------------|
| $3^{-1} \bmod 7$   | $7 = 2 \times 3 + 1$         | $1 = 7 - 2 \times 3$ | $-2 + 7 = 5$         | $3 \times 5 = 15 \equiv 1$ ✓ |
| $1914^{-1} \bmod 899$ | $\gcd = 29 \neq 1$        | -                    | Does not exist       | -                            |

### **Chinese Remainder Theorem**

The **Chinese Remainder Theorem (CRT)** solves a system of congruences. Given positive integers $n_1, n_2, \ldots, n_k$ that are **Pairwise Coprime** (every pair has $\gcd = 1$) and any values $a_1, a_2, \ldots, a_k$, the system:

$$
\begin{aligned}
x &\equiv a_1 \pmod{n_1} \\
x &\equiv a_2 \pmod{n_2} \\
&\ \ \vdots \\
x &\equiv a_k \pmod{n_k}
\end{aligned}
$$

has a **Unique Solution Modulo $N = n_1 n_2 \cdots n_k$**.

- **Chinese Remainder Theorem**:
  - Moduli must be pairwise coprime
  - A solution always exists
  - The solution is unique modulo $N = n_1 n_2 \cdots n_k$

#### **Solving by Substitution**

The idea is to turn one congruence into an equation, then **Substitute It into the Next Congruence**. Each substitution adds one more condition, until all congruences are satisfied.

- **Substitution Method**:
  - Step 1: Take the congruence with the largest modulus, $x \equiv a_i \pmod{n_i}$, and write it as an equation: $x = n_i j + a_i$ for some integer $j$
  - Step 2: Substitute $x$ into the congruence with the next largest modulus
  - Step 3: Solve this new congruence for $j$
  - Step 4: Write $j$ as an equation and substitute it back into $x$
  - Step 5: Repeat until every congruence has been used

#### **CRT Execution for $x \equiv 1 \pmod{3}$, $x \equiv 4 \pmod{5}$, $x \equiv 6 \pmod{7}$**

Moduli $3, 5, 7$ are pairwise coprime, so a unique solution exists modulo $N = 3 \times 5 \times 7 = 105$.

| Step | Action                                  | Working                                                        | Result                    |
|------|-----------------------------------------|----------------------------------------------------------------|---------------------------|
| 1    | Write largest modulus (7) as equation   | $x \equiv 6 \pmod{7}$                                          | $x = 7j + 6$              |
| 2    | Substitute into mod 5                   | $7j + 6 \equiv 4 \pmod{5}$                                     | $2j \equiv 3 \pmod{5}$    |
| 3    | Solve for $j$                           | $2^{-1} \equiv 3 \pmod{5}$, so $j \equiv 3 \times 3 = 9$       | $j \equiv 4 \pmod{5}$     |
| 4    | Write $j$ as equation                   | $j = 5k + 4$                                                   | $j = 5k + 4$              |
| 5    | Substitute into $x$                     | $x = 7(5k + 4) + 6$                                            | $x = 35k + 34$            |
| 6    | Substitute into mod 3                   | $35k + 34 \equiv 1 \pmod{3}$, and $35 \equiv 2$, $34 \equiv 1$ | $2k \equiv 0 \pmod{3}$    |
| 7    | Solve for $k$                           | $\gcd(2, 3) = 1$, so divide by 2                               | $k \equiv 0 \pmod{3}$     |
| 8    | Write $k$ as equation, substitute       | $k = 3l$, so $x = 35(3l) + 34$                                 | $x = 105l + 34$           |

Final result:
- $x = 105l + 34$ for any integer $l$
- Therefore, $x \equiv 34 \pmod{105}$
- Check:
  - $34 \bmod 3 = 1$ ✓
  - $34 \bmod 5 = 4$ ✓
  - $34 \bmod 7 = 6$ ✓

### **Groups, Rings and Fields**

**Groups**, **Rings** and **Fields** are part of **Abstract Algebra**. Abstract algebra studies sets where you can **Combine Two Elements to Get a Third Element** in the same set. The operations are defined by the set's own rules, so they don't have to be ordinary arithmetic.

Each structure builds on the one before it by adding more rules: a ring is a group with extra rules, and a field is a ring with extra rules.

#### **Groups**

A group $\{G, \cdot\}$ is a **Set $G$ with One Binary Operation $\cdot$**. The operation $\cdot$ is generic and could be addition, multiplication or something else. It takes any ordered pair $(a, b)$ in $G$ and gives an element $a \cdot b$, and must satisfy the axioms below.

- **Group Axioms**:
  - **A1 - Closure**: If $a, b \in G$, then $a \cdot b \in G$
  - **A2 - Associative**: $a \cdot (b \cdot c) = (a \cdot b) \cdot c$ for all $a, b, c \in G$
  - **A3 - Identity Element**: There is an $e \in G$ such that $a \cdot e = e \cdot a = a$ for all $a \in G$
  - **A4 - Inverse Element**: For each $a \in G$, there is an $a' \in G$ such that $a \cdot a' = a' \cdot a = e$

- **Abelian Group**: A group that also satisfies
  - **A5 - Commutative**: $a \cdot b = b \cdot a$ for all $a, b \in G$

- **Types of Groups**:
  - **Finite Group**: Has a finite number of elements
  - **Infinite Group**: Has infinitely many elements
  - **Order**: The number of elements in a finite group
  - **Cyclic Group**: Every element is a power $a^k$ of one element $a$, called the **generator**
    - Always abelian
    - Can be finite or infinite

| Example                     | Operation      | Identity | Inverse of $a$ | Abelian? | Finite? |
|-----------------------------|----------------|----------|----------------|----------|---------|
| Integers $\mathbb{Z}$       | Addition       | 0        | $-a$           | Yes      | No      |
| $\{0, 1, \ldots, n - 1\}$   | Addition mod $n$ | 0      | $n - a$        | Yes      | Yes, order $n$ |
| Nonzero reals               | Multiplication | 1        | $1/a$          | Yes      | No      |

#### **Rings**

A ring $\{R, +, \times\}$ is a **Set $R$ with Two Binary Operations**, called **addition** and **multiplication**. Under addition, a ring is always an **Abelian Group**. Multiplication then adds its own axioms.

- **Ring Axioms**:
  - **A1 - A5**: $R$ is an abelian group under addition
  - **M1 - Closure under Multiplication**: If $a, b \in R$, then $ab \in R$
  - **M2 - Associativity of Multiplication**: $a(bc) = (ab)c$ for all $a, b, c \in R$
  - **M3 - Distributive Laws**: $a(b + c) = ab + ac$ and $(a + b)c = ac + bc$ for all $a, b, c \in R$

- **Commutative Ring**: A ring that also satisfies
  - **M4 - Commutativity of Multiplication**: $ab = ba$ for all $a, b \in R$

- **Integral Domain**: A commutative ring that also satisfies
  - **M5 - Multiplicative Identity**: There is an element $1 \in R$ such that $a1 = 1a = a$ for all $a \in R$
  - **M6 - No Zero Divisors**: If $ab = 0$, then $a = 0$ or $b = 0$

| Example                        | Commutative Ring? | Has 1? | No Zero Divisors?            | Integral Domain? |
|--------------------------------|-------------------|--------|------------------------------|------------------|
| Integers $\mathbb{Z}$          | Yes               | Yes    | Yes                          | Yes              |
| $\{0, 1, \ldots, 5\}$ mod 6    | Yes               | Yes    | No: $2 \times 3 \equiv 0 \pmod 6$ | No           |
| $\{0, 1, \ldots, 6\}$ mod 7    | Yes               | Yes    | Yes                          | Yes              |

#### **Fields**

A field $\{F, +, \times\}$ is a **Set $F$ with Addition and Multiplication** that satisfies every group and ring axiom above. A field is always an **Integral Domain**, with one more rule: every nonzero element has a **Multiplicative Inverse**. This means you can always divide by any nonzero element.

- **Field Axioms**:
  - **A1 - A5** and **M1 - M6**: $F$ is an integral domain
  - **M7 - Multiplicative Inverse**: Each $a \in F$ except $0$ has an inverse $a^{-1}$ such that $aa^{-1} = a^{-1}a = 1$

| Example                     | Field? | Reason                                           |
|-----------------------------|--------|--------------------------------------------------|
| Rationals $\mathbb{Q}$, Reals $\mathbb{R}$ | Yes | Every nonzero element has an inverse $1/a$  |
| Integers $\mathbb{Z}$       | No     | $2$ has no integer inverse                       |
| $\{0, 1, \ldots, 6\}$ mod 7 | Yes    | 7 is prime, so every nonzero element has an inverse |
| $\{0, 1, \ldots, 5\}$ mod 6 | No     | 6 is not prime, e.g. $2$ has no inverse mod 6   |

#### **Finite Fields**

**Finite Fields** are the most useful in cryptography. A finite field must have an order of the form $p^n$, where $p$ is **Prime** and $n$ is a **Positive Integer**. It is written as:

$$GF(p^n)$$

**GF** stands for **Galois Field**, named after Évariste Galois, the mathematician who first studied finite fields.

- **Finite Fields in Cryptography**: Order must be $p^n$
  - **$GF(p)$**: when $n = 1$, the integers $\{0, 1, \ldots, p - 1\}$ with arithmetic mod $p$
  - **$GF(2^n)$**: when $p = 2$, works naturally with binary data, used in algorithms like AES

#### **Summary of Structures**

| Structure        | Axioms             | Operations         | Key Property Added                  |
|------------------|--------------------|--------------------|-------------------------------------|
| Group            | A1 - A4            | One                | Can combine and undo                |
| Abelian Group    | A1 - A5            | One                | Order doesn't matter                |
| Ring             | A1 - A5, M1 - M3   | Two                | Adds multiplication                 |
| Commutative Ring | A1 - A5, M1 - M4   | Two                | Multiplication order doesn't matter |
| Integral Domain  | A1 - A5, M1 - M6   | Two                | Has 1, no zero divisors             |
| Field            | A1 - A5, M1 - M7   | Two                | Can divide by any nonzero element   |
