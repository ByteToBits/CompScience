### **Introduction to computer security**

Security is no longer just a strong lock and door. Digital communication has created new ways for breaches to happen, so both sender and receiver must protect information in storage and in transit. The classic problem remains: how can Alice send Bob something that only Bob can open, even if others intercept it along the way?

### **Challenges to Computer Security**

Listing the properties a secure system needs is easy, but **Implementing and Enforcing Them is Hard**. How secure a system is depends on three things: the **Type of Attack**, the **Mechanism Used to Attack**, and the **Service or Functionality Targeted**.

#### **Type of Attack**

Attacks can be grouped by how they work (**Active** or **Passive**) or by where they come from (**Inside** or **Outside** the organisation).

In an active attack, the attacker changes the communication between the sender and receiver. They pretend to be one of the two parties so they can steal sensitive information for money or to cause damage. These are the attacks that usually make the news.

*Masquerade: an example of an active attack*

In a passive attack, the attacker secretly listens in without the sender or receiver knowing. These attacks get less attention, but they can be just as important. The information gathered can be used later to launch an active attack.

*Release of message contents: an example of a passive attack*

- **Active Attacks**: Attacker modifies the communication
  - Attacker pretends to be the sender or receiver
  - Goal is to steal sensitive information or cause damage
  - Often seen as the most dangerous because they make headlines

- **Passive Attacks**: Attacker secretly intercepts the communication
  - Neither the sender nor receiver knows it is happening
  - Collected information can be used for a future active attack
  - Less visible, but just as important to defend against

- **Inside Attacks**: Started by someone inside the organisation
  - Insider already has access to resources
  - Misuses that access in unintended ways

- **Outside Attacks**: Started by someone outside the organisation
  - Usually not connected to the organisation
  - Usually has no prior knowledge of its internals

| Attack Type | Classified By | Attacker               | Victim Aware? | Example                     |
|-------------|---------------|------------------------|---------------|-----------------------------|
| Active      | Method        | Modifies communication | Often yes     | Masquerade                  |
| Passive     | Method        | Only intercepts        | No            | Release of message contents |
| Inside      | Source        | Insider with access    | Varies        | Misusing granted access     |
| Outside     | Source        | Outsider, no knowledge | Varies        | External intrusion          |

#### **Mechanism Used to Attack**

The attack mechanism is **How the Attacker Attacks the System**. The attacker can target any part of the communication between the sender and receiver. There are so many possible combinations, and no two systems are exactly alike, so securing every system is difficult.

- **Possible Targets**:
  - The system itself
  - Specific or installed software on the machine
  - Protocols used by the system or its peripherals
  - The network the systems communicate over

#### **Service or Functionality Targeted**

The attacker's goal is to **Gain an Advantage Over the Target System**. They do this by exploiting a service or feature the system offers.

- **Possible Outcomes**:
  - Access to information or resources
  - Compromised integrity of the data being transmitted

#### **Why Securing a System is Challenging**

Attackers have many options, so defenders first need to understand every way a system could be compromised before they can secure it.

- **Hard to Implement Properties**: Properties like confidentiality and integrity are simple to define but hard to apply to complex systems
- **Need to Understand Mechanisms**: Knowing *how* a feature works is necessary to secure it
- **Attacker vs Defender**: Security is a contest between the attacker's knowledge of how to exploit a system and the defender's knowledge of how to protect it
- **Reactive Mindset**: Users often only think about security after an attack happens
- **Constant Effort Required**: Security needs ongoing monitoring and updates, which is not always possible
- **Usability Trade-off**: Users often see strong security as making systems harder to use

### **Types of Security Attacks**

Security attacks can be classified as **Passive** or **Active**. Each type works differently, has a different **Impact** on the organisation, and leads to different **Threat Consequences**.

#### **Passive Attacks**

In a passive attack, the attacker **Secretly Collects Information** without changing anything. The sender and receiver usually don't know it is happening.

In a release of message contents attack, the attacker reads the information being sent between two parties. They usually do this by exploiting the transmission medium. Traffic analysis is even harder to notice. The attacker doesn't read the messages, but studies the communication patterns. From this, they can guess what the topic is or predict when the next message will be sent.

- **Release of Message Contents**: Attacker reads the information being sent
  - Exploits the transmission medium
  - Obtains information only meant for the receiver

- **Traffic Analysis**: Attacker studies communication patterns
  - Does not need to know the exact contents
  - Can infer the topic of discussion
  - Can predict when the next message will be sent

#### **Active Attacks**

In an active attack, the attacker **Interferes with the Communication or System**. Many active attacks begin with a passive attack to collect the information they need.

- **Masquerade**: Attacker pretends to be the sender or receiver
  - Computers identify each other using an identification metric
  - Attacker only needs to obtain this metric to impersonate someone
  - The metric is usually obtained through a passive attack

- **Replay**: Attacker resends a previously captured message
  - Captures and analyses traffic first
  - Replays the message to trick one party into thinking it came from the other
  - Used to continue the conversation or gain unauthorised access

- **Modification of Messages**: Attacker changes a captured message and sends it on
  - Intercepts the communication, often causing a delay or error
  - Receiver is tricked into giving the attacker unauthorised access

- **Denial of Service (DoS)**: Attacker uses up all of a system's resources
  - The system can no longer handle requests from real users
  - Real users cannot use the service
  - Can be carried out in many different ways

| Attack                      | Type    | What the Attacker Does         | Goal                          |
|-----------------------------|---------|--------------------------------|-------------------------------|
| Release of Message Contents | Passive | Reads transmitted data         | Obtain private information    |
| Traffic Analysis            | Passive | Studies communication patterns | Infer topic or timing         |
| Masquerade                  | Active  | Pretends to be someone else    | Deceive other systems         |
| Replay                      | Active  | Resends an old message         | Continue session or gain access |
| Modification of Messages    | Active  | Alters and resends a message   | Gain unauthorised access      |
| Denial of Service           | Active  | Monopolises system resources   | Make the service unavailable  |

#### **Impact of a Security Attack**

Every attack is unwanted, but some cause more damage than others. Attacks are grouped by **How Badly They Affect the Organisation**.

- **Low Impact**: Limited effect
  - Organisation can still provide its services
  - Some drop in service quality
  - Minor compromise of assets, minor financial loss or minor harm to individuals

- **Medium Impact**: Adverse effect
  - Organisation may still operate
  - Significant drop in service quality
  - Significant compromise of assets, significant financial loss or serious harm to individuals

- **High Impact**: Catastrophic effect
  - Organisation may no longer be able to operate
  - Complete compromise of assets, major financial loss or major harm to individuals

| Impact | Can Still Operate? | Service Quality      | Assets                    | Financial Loss | Harm to Individuals |
|--------|--------------------|----------------------|---------------------------|----------------|---------------------|
| Low    | Yes                | Some degradation     | Slightly compromised      | Minor          | Minor               |
| Medium | Maybe              | Significant drop     | Significantly compromised | Significant    | Serious             |
| High   | Possibly not       | Severe or none       | Completely compromised    | Major          | Major               |

#### **Threat Consequences of Security Attacks**

When an attack happens, it usually leads to one or more of four **Threat Consequences**. Each one threatens a different security property.

- **Unauthorised Disclosure**: Threat to **Confidentiality**
  - Information is seen by someone it wasn't meant for
  - **Exposure**: Sensitive information is deliberately released
  - **Interception**: Data is accessed while being sent from sender to receiver
  - **Interference**: Data is accessed as a by-product of the communication
  - **Intrusion**: Someone gets around the system's security to access data

- **Deception**: Threat to **Integrity**
  - Receiver gets modified information without knowing it
  - **Masquerade**: Gaining access by pretending to be someone else
  - **Falsification**: Changing data sent to an authorised entity without them knowing
  - **Repudiation**: Falsely denying sending a message or performing an action

- **Disruption**: Threat to **Availability**
  - System becomes unavailable to other users
  - **Incapacitation**: Disabling a critical component
  - **Corruption**: Changing key system data or functions so the system behaves badly
  - **Obstruction**: Getting in the way of normal operations

- **Usurpation**: Threat to **Integrity**
  - System is compromised but still running, and users don't notice
  - **Misappropriation**: Taking control of a system resource
  - **Misuse**: Making the system do something that harms its own security

| Threat Consequence      | Property Threatened | Example Attacks                                    |
|-------------------------|---------------------|----------------------------------------------------|
| Unauthorised Disclosure | Confidentiality     | Exposure, Interception, Interference, Intrusion    |
| Deception               | Integrity           | Masquerade, Falsification, Repudiation             |
| Disruption              | Availability        | Incapacitation, Corruption, Obstruction            |
| Usurpation              | Integrity           | Misappropriation, Misuse                           |

In practice, most real attacks combine several of these concepts.

### **OSI Security Architecture**

Securing one system is already hard. Securing many systems across one or more networks is even harder, so organisations need a **Clear Plan** to make sure nothing is missed.

The **Open System Interconnect (OSI) Security Architecture**, also called **Recommendation X.800** (1991), gives a systematic way to secure systems in an organisation. It is an internationally recognised standard, and many vendors make products that follow it. It focuses on three main aspects: **Security Attacks**, **Security Mechanisms** and **Security Services**.

- **Security Attack**: Any action that compromises the security of an organisation's information
  - Includes the passive and active attacks from the previous lesson

- **Security Mechanism**: A process (or device using that process) that deals with attacks
  - Can **detect**, **prevent** or **recover** from an attack

- **Security Service**: A service that improves the security of data processing and information transfer
  - Designed to counter security attacks
  - Uses one or more security mechanisms to work

#### **Security Services**

X.800 divides security services into **Six Categories**. Each one protects a different part of the communication.

Authentication makes sure the communication is genuine. For a single message, the receiver needs to know the sender is real. For an ongoing conversation, both sides must be verified before it starts, and no one should be able to interfere while it is happening.

- **Authentication**: Ensures the communication is genuine
  - **Peer Entity Authentication**: Verifies both peers are who they claim to be
    - Mainly used when the connection is set up, sometimes during it
  - **Data Origin Authentication**: Verifies where the data came from
    - Does not protect against modification or duplication of the data

- **Access Control**: Controls access to systems and resources
  - Requesting system must first be identified
  - Then it must authenticate before it can access the resource

- **Data Confidentiality**: Protects the communication from being sniffed or replayed
  - A listener should not learn anything, including source, destination or frequency

- **Data Integrity**: Protects the data stream between peers
  - Data arrives exactly as sent
  - No modification, duplication, replay or reordering

- **Non-Repudiation**: Makes both peers accountable
  - Neither peer can deny sending a message

- **Availability**: Makes sure the system is available when needed
  - Needs other services working together to achieve it

| Service              | Protects Against                         | Key Idea                        |
|----------------------|------------------------------------------|---------------------------------|
| Authentication       | Impersonation                            | Are you who you say you are?    |
| Access Control       | Unauthorised use of resources            | Are you allowed to use this?    |
| Data Confidentiality | Sniffing, traffic analysis               | Can anyone else read this?      |
| Data Integrity       | Modification, duplication, replay, reordering | Was this changed on the way?    |
| Non-Repudiation      | Denying a message was sent               | Can you deny you sent this?     |
| Availability         | Denial of service                        | Is the system there when needed? |

#### **Security Mechanisms**

X.800 splits security mechanisms into **Two Categories**, based on whether they belong to a specific protocol layer.

**Specific security mechanisms** are built into a particular protocol layer. **Pervasive security mechanisms** are not tied to any layer and apply across the whole system.

- **Specific Security Mechanisms**: Implemented in a specific protocol layer
  - **Encipherment**: Uses mathematical algorithms to turn readable text into unreadable text, which can only be reversed with the key
  - **Digital Signature**: Lets the receiver check where the message came from and that it wasn't changed
  - **Access Control**: Manages access rights to resources
  - **Data Integrity**: Keeps data unchanged
  - **Authentication Exchange**: Confirms an entity's identity by exchanging information
  - **Traffic Padding**: Adds extra bits to the data stream without changing the message, making traffic analysis harder
  - **Routing Control**: Uses secure routes and can switch routes if a breach is detected
  - **Notarisation**: Uses a third party that both sides trust to handle the exchange

- **Pervasive Security Mechanisms**: Not tied to any protocol layer
  - **Trusted Functionality**: Defines criteria that are considered correct and secure
  - **Security Label**: Marks a resource with its security attributes
  - **Event Detection**: Detects security-related events
  - **Security Audit Trail**: Records operations so they can be independently reviewed after a breach
  - **Security Recovery**: Carries out recovery actions when other mechanisms request it

#### **Relationship Between Services and Mechanisms**

Each security service is provided by **One or More Security Mechanisms**.

| Service                       | Encipherment | Digital Signature | Access Control | Data Integrity | Authentication Exchange | Traffic Padding | Routing Control | Notarisation |
|-------------------------------|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| Peer Entity Authentication    | ✓ | ✓ |   |   | ✓ |   |   |   |
| Data Origin Authentication    | ✓ | ✓ |   |   |   |   |   |   |
| Access Control                |   |   | ✓ |   |   |   |   |   |
| Confidentiality               | ✓ |   |   |   |   |   | ✓ |   |
| Traffic Flow Confidentiality  | ✓ |   |   |   |   | ✓ | ✓ |   |

| Non-Repudiation               |   | ✓ |   | ✓ |   |   |   | ✓ |

Many of these mechanisms share one common foundation: **Cryptography**, which will be the main focus going forward.

### **Functional Requirements and Design Principles**

Even with the OSI security architecture, it is still **Impossible to Build a Completely Secure System**. To get as close as possible, government organisations defined a set of **Functional Requirements** and **Design Principles** that their systems must follow.

#### **Functional Requirements**

**FIPS 200** (Minimum Security Requirements for Federal Information and Information Systems) lists the functional requirements for securing federal systems. They can be **Technical** (applied to hardware, software or both) or **Managerial** (how the systems are managed).

- **Access Control**: Only authorised users can access the system and its services
- **Awareness and Training**: All users are trained to handle the security risks of using the system
- **Audit and Accountability**: Activity logs are kept so every user is accountable for their actions
- **Certification, Accreditation and Security Assessments**: Systems are monitored for vulnerabilities, with a plan to fix any found
- **Configuration Management**: Baseline configurations are recorded and used to enforce security
- **Contingency Planning**: Emergency plans exist to restore critical services as quickly as possible
- **Incident Response**: Incidents are monitored, recorded, reported and handled, with plans to detect, contain and recover
- **Maintenance**: Systems are regularly assessed and maintained or upgraded when needed
- **Media Protection**: Both hard copy and digital media are protected
- **Physical and Environmental Protection**: Physical access is limited, and areas have proper environmental controls
- **Planning**: Security plans cover how to set up, run and maintain security for all systems
- **Personnel Security**: Staff in charge of critical systems must meet key criteria
- **Risk Assessment**: Risk assessments are done regularly on key systems and processes
- **Systems and Services Acquisition**: Resources (internal or external) are in place to protect the systems
- **System and Communications Protection**: All communications are monitored, controlled and protected
- **System and Information Integrity**: Integrity is maintained at all times, with a response plan if it is compromised

#### **Design Principles**

Systems are complex and varied, so these widely accepted **Design Principles** guide how security mechanisms are built.

- **Economy of Mechanism**: Keep designs small and simple
  - Simple designs are easier to test and verify
- **Fail-Safe Defaults**: Deny access by default
  - Access is granted only by explicit permission
- **Complete Mediation**: Check every request every time
  - Don't rely on cached or previously granted access
- **Open Design**: The design should be public, not secret
- **Separation of Privilege**: Access depends on multiple attributes together
- **Least Privilege**: Give users and processes only the minimum privileges they need
- **Least Common Mechanism**: Users should share as little functionality as possible
  - Fewer shared paths means fewer ways for problems to spread
- **Psychological Acceptability**: Security shouldn't get in the way of users' work, while still being secure
- **Isolation**: Keep systems separated based on what they are used for
- **Encapsulation**: Use object-oriented techniques so functionality is self-contained
- **Modularity**: Build security as separate functional modules
- **Layering**: Apply security in layers covering people, technology and operations
- **Least Astonishment**: The system should respond in ways that don't surprise the user

| Principle                   | Key Idea                              |
|-----------------------------|---------------------------------------|
| Economy of Mechanism        | Keep it simple                        |
| Fail-Safe Defaults          | Deny unless allowed                   |
| Complete Mediation          | Check every request                   |
| Open Design                 | No security through secrecy           |
| Separation of Privilege     | Multiple conditions for access        |
| Least Privilege             | Only what is needed                   |
| Least Common Mechanism      | Minimise shared functionality         |
| Psychological Acceptability | Easy for users                        |
| Isolation                   | Separate systems by purpose           |
| Encapsulation               | Self-contained functionality          |
| Modularity                  | Separate modules                      |
| Layering                    | Defence in depth                      |
| Least Astonishment          | No surprises for users                |

These requirements and principles are theoretical, but they are the **Accepted Practices** behind how modern security mechanisms are designed.

### **A Model for Network Security**

Protecting stored information matters, but the most common case is **Information Being Sent from One Entity to Another**. Networks handle this in many ways, and the job is to put the **Right Mechanisms in the Right Places** so the communication channel is secure.

Information sent over a network can always be intercepted by an adversary. This is just how transmission works, and the medium can't be changed. So instead of changing the medium, we **Change the Information** itself using **Cryptography**, which is the foundation that many other security mechanisms are built on.

Encryption only works if **Both Parties Have the Key**. This creates a problem: how do we safely share the key? If both sides could meet in person to exchange it, they could just share the message directly and wouldn't need encryption at all.

- **Information in Transit**: Can always be intercepted
  - Interception is part of how transmission works
  - The transmission medium cannot be changed

- **Solution**: Change the information instead of the medium
  - Use cryptography to make intercepted information unreadable
  - Cryptography is the basis for many other security mechanisms

- **Key Distribution Problem**: Both parties need the same key
  - How do you send the key securely?
  - If you can meet in person to share the key, why encrypt at all?

Before learning encryption techniques, we first need to cover the **Mathematical Principles** that cryptography is built on.
